# 消融实验，CL的作用，memory的作用

```python
import json
import matplotlib.pyplot as plt
import numpy as np

# 加载数据
with open('test_than_train_future_test_results.json', 'r') as f:
    baseline_data = json.load(f)

with open('CL_unk_test_than_train_future_month.json', 'r') as f:
    cl_data = json.load(f)

with open('CL_unk_no_memory_test_than_train_future_month.json', 'r') as f:
    no_mem_data = json.load(f)

months = baseline_data['month']
baseline_precision = baseline_data['precision']
cl_precision = cl_data['precision']
no_mem_precision = no_mem_data['precision']
baseline_recall = baseline_data['recall']
cl_recall = cl_data['recall']
no_mem_recall = no_mem_data['recall']
baseline_f1 = baseline_data['f1']
cl_f1 = cl_data['f1']
no_mem_f1 = no_mem_data['f1']
```

```python
# 绘制3个子图：Precision、Recall、F1
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 子图1: Precision
axes[0].plot(months, baseline_precision, 'o-', label='Baseline', color='steelblue', linewidth=2, markersize=5)
axes[0].plot(months, cl_precision, 's-', label='CL', color='coral', linewidth=2, markersize=5)
axes[0].plot(months, no_mem_precision, 'd-', label='No Memory CL', color='seagreen', linewidth=2, markersize=5)
axes[0].set_xlabel('Month', fontsize=11)
axes[0].set_ylabel('Precision', fontsize=11)
axes[0].set_title('Precision', fontsize=13, fontweight='bold')
axes[0].legend()
axes[0].set_ylim(0, 1.1)
axes[0].grid(alpha=0.3)
axes[0].tick_params(axis='x', rotation=45)

# 子图2: Recall
axes[1].plot(months, baseline_recall, 'o-', label='Baseline', color='steelblue', linewidth=2, markersize=5)
axes[1].plot(months, cl_recall, 's-', label='CL', color='coral', linewidth=2, markersize=5)
axes[1].plot(months, no_mem_recall, 'd-', label='No Memory CL', color='seagreen', linewidth=2, markersize=5)
axes[1].set_xlabel('Month', fontsize=11)
axes[1].set_ylabel('Recall', fontsize=11)
axes[1].set_title('Recall', fontsize=13, fontweight='bold')
axes[1].legend()
axes[1].set_ylim(0, 1.1)
axes[1].grid(alpha=0.3)
axes[1].tick_params(axis='x', rotation=45)

# 子图3: F1
axes[2].plot(months, baseline_f1, 'o-', label='Baseline', color='steelblue', linewidth=2, markersize=5)
axes[2].plot(months, cl_f1, 's-', label='CL', color='coral', linewidth=2, markersize=5)
axes[2].plot(months, no_mem_f1, 'd-', label='No Memory CL', color='seagreen', linewidth=2, markersize=5)

axes[2].set_xlabel('Month', fontsize=11)
axes[2].set_ylabel('F1 Score', fontsize=11)
axes[2].set_title('F1 Score', fontsize=13, fontweight='bold')
axes[2].legend()
axes[2].set_ylim(0, 1.1)
axes[2].grid(alpha=0.3)
axes[2].tick_params(axis='x', rotation=45)

plt.suptitle('Baseline vs CL on Future Months (2023-03 ~ 2024-12)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# 统计对比
baseline_f1_arr = np.array(baseline_f1)
cl_f1_arr = np.array(cl_f1)
no_mem_f1_arr = np.array(no_mem_f1)

print("="*50)
print("统计对比")
print("="*50)
print(f"Baseline - Mean Precision: {np.mean(baseline_precision):.4f}, Mean Recall: {np.mean(baseline_recall):.4f}, Mean F1: {np.mean(baseline_f1_arr):.4f}")
print(f"CL       - Mean Precision: {np.mean(cl_precision):.4f}, Mean Recall: {np.mean(cl_recall):.4f}, Mean F1: {np.mean(cl_f1_arr):.4f}")
print(f"No Memory CL - Mean Precision: {np.mean(no_mem_precision):.4f}, Mean Recall: {np.mean(no_mem_recall):.4f}, Mean F1: {np.mean(no_mem_f1_arr):.4f}")

improved = np.sum(cl_f1_arr > baseline_f1_arr)
degraded = np.sum(cl_f1_arr < baseline_f1_arr)
same = np.sum(cl_f1_arr == baseline_f1_arr)
print(f"CL相比Baseline提升的月份数: {improved}, 下降: {degraded}, 相同: {same}")
# CL相比no_memory提升的月份
improved_no_mem = np.sum(cl_f1_arr > no_mem_f1_arr)
degraded_no_mem = np.sum(cl_f1_arr < no_mem_f1_arr)
same_no_mem = np.sum(cl_f1_arr == no_mem_f1_arr)
print(f"CL相比No Memory CL提升的月份数: {improved_no_mem}, 下降: {degraded_no_mem}, 相同: {same_no_mem}")
```

```python
# 加载数据
with open('CL_unk_test_than_train_seen_month.json', 'r') as f:
    CL_seen_month = json.load(f)

with open('CL_unk_no_memory_test_than_train_seen_month.json', 'r') as f:
    no_mem_seen_month = json.load(f)

months = CL_seen_month['month']
CL_seen_month_precision = CL_seen_month['precision']
no_mem_seen_month_precision = no_mem_seen_month['precision']
CL_seen_month_recall = CL_seen_month['recall']
no_mem_seen_month_recall = no_mem_seen_month['recall']
CL_seen_month_f1 = CL_seen_month['f1']
no_mem_seen_month_f1 = no_mem_seen_month['f1']
```

```python
# 绘制3个子图：Precision、Recall、F1
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 子图1: Precision
axes[0].plot(months, CL_seen_month_precision, 'o-', label='Baseline', color='steelblue', linewidth=2, markersize=5)
axes[0].plot(months, no_mem_seen_month_precision, 's-', label='CL', color='coral', linewidth=2, markersize=5)
axes[0].set_xlabel('Month', fontsize=11)
axes[0].set_ylabel('Precision', fontsize=11)
axes[0].set_title('Precision', fontsize=13, fontweight='bold')
axes[0].legend()
axes[0].set_ylim(0, 1.1)
axes[0].grid(alpha=0.3)
axes[0].tick_params(axis='x', rotation=45)

# 子图2: Recall
axes[1].plot(months, CL_seen_month_recall, 's-', label='CL', color='coral', linewidth=2, markersize=5)
axes[1].plot(months, no_mem_seen_month_recall, 'd-', label='No Memory CL', color='seagreen', linewidth=2, markersize=5)
axes[1].set_xlabel('Month', fontsize=11)
axes[1].set_ylabel('Recall', fontsize=11)
axes[1].set_title('Recall', fontsize=13, fontweight='bold')
axes[1].legend()
axes[1].set_ylim(0, 1.1)
axes[1].grid(alpha=0.3)
axes[1].tick_params(axis='x', rotation=45)

# 子图3: F1
axes[2].plot(months, CL_seen_month_f1, 's-', label='CL', color='coral', linewidth=2, markersize=5)
axes[2].plot(months, no_mem_seen_month_f1, 'd-', label='No Memory CL', color='seagreen', linewidth=2, markersize=5)

axes[2].set_xlabel('Month', fontsize=11)
axes[2].set_ylabel('F1 Score', fontsize=11)
axes[2].set_title('F1 Score', fontsize=13, fontweight='bold')
axes[2].legend()
axes[2].set_ylim(0, 1.1)
axes[2].grid(alpha=0.3)
axes[2].tick_params(axis='x', rotation=45)

plt.suptitle('CL w/o memory on Seen Months', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# 统计对比
CL_seen_month_f1 = np.array(CL_seen_month_f1)
no_mem_seen_month_f1 = np.array(no_mem_seen_month_f1)

print("="*50)
print("统计对比")
print("="*50)
print(f"CL       - Mean Precision: {np.mean(CL_seen_month_precision):.4f}, Mean Recall: {np.mean(CL_seen_month_recall):.4f}, Mean F1: {np.mean(CL_seen_month_f1):.4f}")
print(f"No Memory CL - Mean Precision: {np.mean(no_mem_seen_month_precision):.4f}, Mean Recall: {np.mean(no_mem_seen_month_recall):.4f}, Mean F1: {np.mean(no_mem_seen_month_f1):.4f}")


# CL相比no_memory提升的月份
improved_no_mem = np.sum(CL_seen_month_f1 > no_mem_seen_month_f1)
degraded_no_mem = np.sum(CL_seen_month_f1 < no_mem_seen_month_f1)
same_no_mem = np.sum(CL_seen_month_f1 == no_mem_seen_month_f1)
print(f"CL相比No Memory CL提升的月份数: {improved_no_mem}, 下降: {degraded_no_mem}, 相同: {same_no_mem}")
```
